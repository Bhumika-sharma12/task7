# importing neccessary librairies

```python
import pymysql
import pandas as pd
import matplotlib.pyplot as plt
import pymysql
import sqlalchemy
```

# creating connection

```python
con_obj=sqlalchemy.create_engine('mysql+pymysql://root:root@localhost:3307/orders')
```

# creating dataframe

```python
df=pd.read_sql('select * from sales_orders',con=con_obj)
df
```

```python
df.columns
```

# select query

```python
df2=pd.read_sql('SELECT product_name, SUM(quantity) AS total_qty, SUM(revenue) AS revenue FROM sales_orders GROUP BY product_name',con=con_obj)
df2
```

# graph

```python
df2.plot(kind='bar', x='product_name', y='revenue')
plt.savefig("sales_chart.png")
```
